```python
# Simple Mode Imputation
# Education Level
df['Educational_level']=df['Educational_level'].fillna(df['Educational_level'].mode()[0])
```

```python
df['Educational_level'].isnull().sum()
```

```python
# Vehicle driver relation
df['Vehicle_driver_relation'] = df['Vehicle_driver_relation'].fillna(df['Vehicle_driver_relation'].mode()[0])
```

```python
df['Vehicle_driver_relation'].isnull().sum()
```

```python
# Rule Based If else(loc[])
# Owner of vehicle
# if driver relation is Owner
df.loc[
    (df['Owner_of_vehicle'].isnull()) &
    (df['Vehicle_driver_relation'] == 'Owner'),
    'Owner_of_vehicle'
] = 'Owner'

# if driver relation is Employee
df.loc[
    (df['Owner_of_vehicle'].isnull()) &
    (df['Vehicle_driver_relation'] == 'Employee'),
    'Owner_of_vehicle'
] = 'Company'

# if relation is Other
df.loc[
    (df['Owner_of_vehicle'].isnull()) &
    (df['Vehicle_driver_relation'] == 'Other'),
    'Owner_of_vehicle'
] = 'Other'

# remaining missing values
df['Owner_of_vehicle'] = df['Owner_of_vehicle'].fillna('Unknown')
```

```python
df['Owner_of_vehicle'].isnull().sum()
```

```python
# Group Based Mode Imputation
# Driving Experience
# fiiling missing driving experience based on age group
df['Driving_experience'] = df.groupby('Age_band_of_driver')['Driving_experience'] \
    .transform(lambda x: x.fillna(x.mode()[0]))
```

```python
df['Driving_experience'].isnull().sum()
```

```python
# Context_Based Imputation
# Road Surface Type
# fiiling based on weather condition
df['Road_surface_type'] = df.groupby('Weather_conditions')['Road_surface_type'] \
    .transform(lambda x: x.fillna(x.mode()[0]))
```

```python
df['Road_surface_type'].isnull().sum()
```

```python
# Mapping + fallback Conditional Filling
junction_map = {
    'Residential areas': 'Cross Road',
    'Office areas': 'T-Junction',
    ' Industrial areas': 'Roundabout',
    '  Market areas': 'Y-Junction',
    ' Church areas': 'T-Junction',
    ' Hospital areas': 'T-Junction',
    'School areas': 'Cross Road',
    'Rural village areas': 'No junction',
    ' Outside rural areas': 'No junction',
    'Recreational areas': 'Roundabout',
    '  Recreational areas': 'Roundabout',
    'Other': 'Unknown',
    'Unknown': 'Unknown',
    'Rural village areasOffice areas': 'T-Junction'
}

# Step 1: Fill using mapping
df['Types_of_Junction'] = df['Types_of_Junction'].fillna(
    df['Area_accident_occured'].map(junction_map)
)

# Step 2: Fill rows where Area_accident_occured itself is missing
df.loc[
    df['Types_of_Junction'].isnull(),
    'Types_of_Junction'
] = 'Unknown'
```

```python
df['Types_of_Junction'].isnull().sum()
```

```python
# Custom Function with apply()
# vehicle movement
def fill_vehicle_movement(row):
    if pd.isnull(row['Vehicle_movement']):
        if row['Type_of_collision'] == 'Collision with vehicles':
            return 'Moving Forward'
        elif row['Type_of_collision'] == 'Collision with pedestrians':
            return 'Going Straight'
        else:
            return 'Unknown'
    return row['Vehicle_movement']
#apply function
df['Vehicle_movement'] = df.apply(fill_vehicle_movement, axis=1)
```

```python
df['Vehicle_movement'].isnull().sum()
```

```python
# Creating Unknown category
#defect of vehicle
df['Defect_of_vehicle']=df['Defect_of_vehicle'].fillna('Unkown')
```

```python
df['Defect_of_vehicle'].isnull().sum()
```

```python
# Descriptive Unknown category
# Work of casuality
df['Work_of_casuality']=df['Work_of_casuality'].fillna('Occupation not recorded')
```

```python
df['Work_of_casuality'].isnull().sum()
```

```python
# Health Safe Unkown Replacement
# fitness of casuality
df['Fitness_of_casuality']=df['Fitness_of_casuality'].fillna('Unkown fitness status')
```

```python
df['Fitness_of_casuality'].isnull().sum()
```

```python
# Remaining Low-Missing Columns Using Mode
remaining_mode_cols = [
    'Type_of_vehicle',
    'Service_year_of_vehicle',
    'Area_accident_occured',
    'Lanes_or_Medians',
    'Road_allignment',
    'Type_of_collision'
]
for col in remaining_mode_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
df[remaining_mode_cols].isnull().sum()
```

```python
# Final missing values after cleaning
df.isnull().sum()
```

```python
# Converting time column to datetime
df['Time'] = pd.to_datetime(
    df['Time'].astype(str).str.strip(),
    format='%H:%M:%S',
    errors='coerce'
)
```

```python
#Convert remaining object columns to category
cat_cols = df.select_dtypes(include='object').columns
df[cat_cols] = df[cat_cols].astype('category')
```

```python
#verify
df.info()
```

```python
# =====================================================
# OUTLIER DETECTION
# =====================================================

num_cols = ['Number_of_vehicles_involved', 'Number_of_casualties']
# Dictionary to store IQR limits
limits = {}
print("Outliers Before Handling")
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
# Store limits for later verification
    limits[col] = (lower, upper)
# Count outliers
    outliers = ((df[col] < lower) | (df[col] > upper)).sum()
    print(f"{col}: {outliers} outliers")
```

```python
# OUTLIER HANDLING USING IQR CAPPING
for col in num_cols:
    lower, upper = limits[col]
    # Cap extreme values
    df[col] = df[col].clip(lower, upper)
print("Outlier handling completed")
```

```python
# OUTLIER VERIFICATION
print("\nOutliers After Handling")
for col in num_cols:
    lower, upper = limits[col]
    remaining = ((df[col] < lower) | (df[col] > upper)).sum()
    print(f"{col}: {remaining} outliers remaining")
```

```python
# FEATURE ENGINEERING
# 1. Extract accident hour from Time column
df['Accident_Hour'] = df['Time'].dt.hour

# 2. Create time period feature
def get_time_period(hour):
    if pd.isna(hour):
        return 'Unknown'
    elif 5 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 21:
        return 'Evening'
    else:
        return 'Night'

df['Time_Period'] = df['Accident_Hour'].apply(get_time_period)
```

```python
# 3. Create weather risk indicator
df['High_Risk_Weather'] = np.where(
    df['Weather_conditions'].isin(['Raining', 'Fog or mist', 'Snow']),
    1,
    0
)
```

```python
# 4. Categorize number of vehicles involved
def vehicle_category(v):
    if v <= 1:
        return 'Low'
    elif v <= 3:
        return 'Medium'
    elif v <= 5:
        return 'High'
    else:
        return 'Very High'

df['Vehicles_Category'] = df['Number_of_vehicles_involved'].apply(vehicle_category)
```

```python
# 5. Categorize number of casualties
def casualty_category(c):
    if c == 0:
        return 'No Casualty'
    elif c <= 2:
        return 'Low'
    elif c <= 5:
        return 'Medium'
    else:
        return 'High'

df['Casualties_Category'] = df['Number_of_casualties'].apply(casualty_category)
```

```python
# FEATURE SELECTION
# Drop less useful columns
drop_cols = [
    'Time',                    # replaced by Accident_Hour
    'Pedestrian_movement',
    'Work_of_casuality',
    'Fitness_of_casuality'
]
# Create final selected dataset
df_selected = df.drop(columns=drop_cols)
```
